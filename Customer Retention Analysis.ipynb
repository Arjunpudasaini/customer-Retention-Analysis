Data cleaning

```python
# Display rows that contain missing values
df[df.isnull().any(axis=1)].head(20)
```

```python
# Remove rows with missing customer and order date information

df = df.dropna(subset=['custid', 'created', 'firstorder', 'lastorder'])
```

```python
# Convert firstorder and lastorder into datetime format

df['firstorder'] = pd.to_datetime(df['firstorder'], errors='coerce')
df['lastorder'] = pd.to_datetime(df['lastorder'], errors='coerce')
```

```python
# Check the data types of date columns

df[['created', 'firstorder', 'lastorder']].dtypes
```

```python
# Check missing dates after date conversion

df[['firstorder', 'lastorder']].isnull().sum()
```

```python
# Remove rows with invalid firstorder and lastorder dates

df = df.dropna(subset=['firstorder', 'lastorder'])
```

```python
# Check missing values after date cleaning

df[['firstorder', 'lastorder']].isnull().sum()
```

```python
df.info()
```

Feature Engineering

```python
# Create a service usage score

df['service_score'] = df['paperless'] + df['refill'] + df['doorstep']
```

```python
# Check the new feature

df['service_score'].describe()
```

```python
# Calculate the estimated number of emails opened

df['estimated_emails_opened'] = df['esent'] * df['eopenrate']
```

```python
# Check the new feature

df['estimated_emails_opened'].describe()
```

```python
# Encode categorical fields (favday, city)

df_encoded = pd.get_dummies(df, columns=['favday', 'city'], drop_first=True)
df_encoded.head()
```

```python
# Derive customer tenure (in days) between first and last order

df['customer_tenure'] = (df['lastorder'] - df['firstorder']).dt.days
df['customer_tenure'].describe()
```

Outlier Detection & Treatmen

```python
# Check outliers in important numerical columns

sns.boxplot(data=df[['esent', 'eopenrate', 'eclickrate', 'avgorder', 'ordfreq']])
plt.xticks(rotation=45)
plt.show()
```

```python
# Detect outliers in avgorder using the IQR method

Q1 = df['avgorder'].quantile(0.25)
Q3 = df['avgorder'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['avgorder'] < lower) | (df['avgorder'] > upper)]

print("Number of outliers:", len(outliers))
print("Lower bound:", lower)
print("Upper bound:", upper)
```

1,741 outliers were found in avgorder, but they were kept because they may be real customers.

```python
#Find outliers in ordfreq

Q1 = df['ordfreq'].quantile(0.25)
Q3 = df['ordfreq'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['ordfreq'] < lower) | (df['ordfreq'] > upper)]

print("Outliers:", len(outliers))
print("Lower:", lower)
print("Upper:", upper)
```

3,723 outliers were found in ordfreq

```python
# Find outliers in esent

Q1 = df['esent'].quantile(0.25)
Q3 = df['esent'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['esent'] < lower) | (df['esent'] > upper)]

print("Outliers:", len(outliers))
print("Lower:", lower)
print("Upper:", upper)
```

12 outliers were found in esent, but they were kept because they may be real values

**EDA (Exploratory Data Analysis)**

```python
# Check customer retention distribution
sns.countplot(data=df, x='retained')
plt.title('Customer Retention')
plt.show()
```

Most customers were retained, while fewer customers were not retained.

```python
# Compare average order value by retention

avg_order = df.groupby('retained')['avgorder'].mean()

bars = plt.bar(['Not Retained', 'Retained'], avg_order)

plt.title('Average Order Value by Retention')
plt.xlabel('Customer Retention')
plt.ylabel('Average Order Value')

plt.bar_label(bars, fmt='%.2f')

plt.show()
```

Most retained customers have a higher average order value.

```python
# Create clear retention labels

df['retention_status'] = df['retained'].map({
    0: 'Non-retained',
    1: 'Retained'
})
```

```python
# Compare average order frequency with customer retention

sns.barplot(data=df, x='retained', y='ordfreq', estimator='mean')
plt.title('Average Order Frequency by Retention')
plt.xlabel('Customer Retention')
plt.ylabel('Average Order Frequency')
plt.show()
```

Retained customers usually have higher order frequency.

```python
#Compare email open rate with customer retention

sns.boxplot(data=df, x='retained', y='eopenrate')
plt.title('Email Open Rate by Retention')
plt.show()
```

Customers who were not retained had a higher email open rate.

```python
#Compare email click rate with customer retention
sns.violinplot(data=df, x='retained', y='eclickrate')
plt.title('Email Click Rate by Retention')
plt.show()
```

Retained customers usually have a higher email click rate.

```python
#Check retention by favorite day

sns.countplot(data=df, x='favday', hue='retained')
plt.title('Customer Retention by Favorite Day')
plt.xticks(rotation=45)
plt.show()
```

Monday had the highest number of retained customers, followed by Tuesday.

```python
# Check retention by city
sns.countplot(data=df, x='city', hue='retained')
plt.title('Customer Retention by City')
plt.xticks(rotation=45)
plt.show()
```

BOM city has the highest number of both retained and non-retained customers.

```python
#Check correlations between important features

corr = df[['retained', 'esent', 'eopenrate', 'eclickrate', 'avgorder', 'ordfreq', 'service_score']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Email sent has the strongest relationship with customer retention, while the other features have very weak relationships

```python
#Check customer retention over time
df['created_year'] = df['created'].dt.year
sns.countplot(data=df, x='created_year', hue='retention_status')
plt.title('Customer Retention Over Time')
plt.xticks(rotation=45)
plt.show()
```

Customer retention was lower in 2008 and highest in 2013.
